```python
import pandas as pd
df=pd.read_csv('/content/car_price_performance.csv')
```

```python
df
```

```python
df.columns
```

```python
nullpricedata=df[df['Price'].isna()]
nullpricedata
```

```python
brand_avg_prices = df.groupby("Brand")["Price"].mean().sort_values()
brand_avg_prices
```

```python
d = {}
count = 1

for i in brand_avg_prices.index:
    d[i] = count
    count += 1

print(d)
```

```python
x = []

for j in df["Brand"]:
    y = d[j]
    x.append(y)
df["Encoded Brand"]=x
df
```

```python
scaledhp=(df["Horsepower"]-df["Horsepower"].min())/(df["Horsepower"].max()-df["Horsepower"].min())
scaledhp
```

```python
df['scaled horsepower'] =scaledhp
df
```

```python
scaledfe=(df["Fuel Efficiency"]-df["Fuel Efficiency"].min())/(df["Fuel Efficiency"].max()-df["Fuel Efficiency"].min())
scaledfe
```

```python
df['scaled fuel efficiency'] =scaledfe
df
```

```python
scaledbrand=(df["Encoded Brand"]-df["Encoded Brand"].min())/(df["Encoded Brand"].max()-df["Encoded Brand"].min())
scaledbrand
```

```python
df['scaled brand'] =scaledbrand
df
```

```python
df['Performance Index'] = ((df['Horsepower'] * 0.7) + (df['Fuel Efficiency'] * 0.3)) / df['Price']
df
```

```python
df.to_csv('full_car_performance_analysis.csv', index=False)

from google.colab import files
files.download('full_car_performance_analysis.csv')
```
